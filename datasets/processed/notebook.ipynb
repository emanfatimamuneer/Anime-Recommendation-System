```python
data_frame = df[["anime_id", "title","score","rank", "synopsis","type", "image_url", "genre","start_date","end_date","episodes"]]
```

```python
data_frame = data_frame.dropna(subset=["start_date", "end_date", "episodes"])
```

```python
data_frame["genre"] = data_frame["genre"].apply(lambda x: [x] if pd.notna(x) else [])
```

```python
print(data_frame.isnull().sum())
```

```python
data_frame.drop_duplicates(inplace=True)   
```

```python
data_frame.duplicated().sum()
```

```python
import re

def clean_genre(text):
    if pd.isna(text):
        return text

    # Remove prefixes like Theme:: or Demographic::
    text = re.sub(r'^[A-Za-z]+::', '', text)

    # Remove duplicated text
    text = re.sub(r'^(.+?)\s+\1$', r'\1', text)

    return text.strip()

data_frame["genre"] = data_frame["genre"].apply(clean_genre)
```

```python
data_frame["synopsis"] = data_frame["synopsis"].apply(
    lambda x: [x] if pd.notna(x) else []
)
```

```python
data_frame["tags"] = data_frame["genre"] + data_frame["synopsis"]
```

```python
data_frame["tags"] = data_frame["tags"].apply(lambda x: " ".join(x))
```

```python
new_data_frame = data_frame[["anime_id", "title","score","rank","type", "image_url","start_date","end_date","episodes","tags"]]
```

```python
new_data_frame.to_csv("processed_data.csv", index=False)
```
