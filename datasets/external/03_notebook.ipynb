```python
import pandas as pd
import pickle
import numpy as np
from pathlib import Path
```

```python
BASE_DIR = Path.cwd().parent.parent

DATA_PATH = BASE_DIR / "datasets" / "processed" / "merged.csv"
SIMILARITY_PATH = BASE_DIR / "datasets" / "external" / "similarity.pkl"

df = pd.read_csv(DATA_PATH)

with open(SIMILARITY_PATH, "rb") as file:
    similarity = pickle.load(file)
```

```python
print(df.shape)
print(similarity.shape)
```

```python
TOP_N = 10

recommendation_map = {}

for anime_index in range(len(df)):

    distances = similarity[anime_index]

    # Get top N + 1
    top_indices = np.argpartition(
        distances,
        -TOP_N - 1
    )[-TOP_N - 1:]

    # Sort selected indices by similarity
    top_indices = top_indices[
        np.argsort(distances[top_indices])[::-1]
    ]

    # Remove the anime itself
    top_indices = [
        index
        for index in top_indices
        if index != anime_index
    ][:TOP_N]

    title = df.iloc[anime_index]["title"]

    recommendation_map[title] = [
        df.iloc[index]["title"]
        for index in top_indices
    ]
```

```python
list(recommendation_map.items())[:3]
```

```python
OUTPUT_PATH = BASE_DIR / "datasets" / "processed" / "recommendations.pkl"

with open(OUTPUT_PATH, "wb") as file:
    pickle.dump(recommendation_map, file)
```

```python
print(
    OUTPUT_PATH,
    OUTPUT_PATH.stat().st_size / (1024 ** 2),
    "MB"
)
```
